```python
#week-3
# Create a dataset with missing values, duplicates, and an outlier
import pandas as pd 
import numpy as np
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David',
             'Alice', 'Emma', 'Frank'],
    'Age': [20, 22, np.nan, 23, 20, 21, 120],
    'Marks': [85, 72, 90, np.nan, 85, 88, 65]
}
print("Original DataSet : ")
df = pd.DataFrame(data)
print(df)
print("Missing Values : ")
print(df.isnull().sum())
print("Fill the null values :")
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
print(df)
print("Duplicates rows:")
print(df.duplicated().sum())
df = df.drop_duplicates()
print("Remove duplicate row :")
print(df)
print("OutLiers : ")
q1 = df["Age"].quantile(0.25)
q3 = df["Age"].quantile(0.75)
iqr = q3 - q1
low = q1 - 1.5*iqr
up = q3 + 1.5*iqr
outliers = df[(df["Age"] < low) | (df["Age"]>up)]
print(outliers)

median_age = df['Age'].median()

df.loc[(df['Age'] < low) | (df['Age'] > up),'Age'] = median_age

print("Dataset after handling outliers:")
print(df)
print("Final Cleaned Dataset:")
print(df)

print("\nMissing values remaining:")
print(df.isnull().sum())

print("\nDuplicate rows remaining:")
print(df.duplicated().sum())
```

```python
#week - 4

import pandas as pd

# Create a sample dataset
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Emma'],
    'Age': [20, 22, 21, 23, 20],
    'Marks': [85, 72, 90, 65, 88],
    'Department': ['CSE', 'ECE', 'CSE', 'EEE', 'ECE']
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

print("Descriptive Statistics:")
print(df.describe())

print("Mean Marks:", df['Marks'].mean())
print("Median Marks:", df['Marks'].median())
print("Mode Marks:", df['Marks'].mode().tolist())
print("Minimum Marks:", df['Marks'].min())
print("Maximum Marks:", df['Marks'].max())
print("Standard Deviation:", df['Marks'].std())

print("Department Frequency:")
print(df['Department'].value_counts())

print("\nUnique Departments:")
print(df['Department'].unique())

print("\nNumber of Unique Departments:")
print(df['Department'].nunique())

print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nStatistical Summary:")
print(df.describe())

print("\nDepartment-wise Average Marks:")
print(df.groupby('Department')['Marks'].mean())
```

```python
#week - 5

import pandas as pd
import numpy as np

# Create a sample dataset
data = {
    'Study_Hours': [1, 2, 3, 4, 5],
    'Marks': [40, 50, 60, 70, 80],
    'Sleep_Hours': [8, 7, 6, 5, 4]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)

correlation = df.corr()

print("Correlation Matrix:")
print(correlation)

covariance = df.cov()

print("Covariance Matrix:")
print(covariance)

corr_value = df['Study_Hours'].corr(df['Marks'])
cov_value = df['Study_Hours'].cov(df['Marks'])

print("Correlation between Study Hours and Marks:", corr_value)
print("Covariance between Study Hours and Marks:", cov_value)

corr_matrix = np.corrcoef(
    df['Study_Hours'],
    df['Marks']
)

print("Correlation using NumPy:")
print(corr_matrix)
```

```python
#week - 6


import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the employee dataset
df = pd.read_csv("employee.csv")

# 1. Display dataset
print("First five records:")
print(df.head())

print("\nDataset shape:")
print(df.shape)

print("\nData types:")
print(df.dtypes)

# 2. Descriptive statistics
print("\nStatistical Summary:")
print(df.describe())

# 3. Check missing values and duplicates
print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

# 4. Department distribution
print("\nDepartment Count:")
print(df['Department'].value_counts())

# 5. Histogram of employee ages
plt.figure(figsize=(7, 4))
sns.histplot(data=df, x='Age', bins=6, kde=True)
plt.title("Distribution of Employee Age")
plt.show()

# 6. Salary distribution by department
plt.figure(figsize=(9, 4))
sns.barplot(data=df, x='Department', y='Salary')
plt.title("Salary by Department")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 7. Experience versus salary
plt.figure(figsize=(7, 4))
sns.scatterplot(data=df, x='Experience', y='Salary')
plt.title("Experience vs Salary")
plt.show()

# 8. Correlation heatmap
plt.figure(figsize=(7, 5))
corr = df[['Age', 'Salary', 'Experience']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Employee Data Correlation")
plt.show()

# 9. Department-wise average salary
print("\nAverage Salary by Department:")
print(df.groupby('Department')['Salary'].mean())

```

```python
#week-7

import pandas as pd
import matplotlib.pyplot as plt

# Load the employee dataset
df = pd.read_csv("employee.csv")

# Display the dataset
print("Employee Dataset:")
print(df.head())

# 1. Line Plot: Employee ID vs Salary
plt.figure(figsize=(8, 5))
plt.plot(df['Emp_ID'], df['Salary'], marker='o')
plt.title("Employee ID vs Salary")
plt.xlabel("Employee ID")
plt.ylabel("Salary")
plt.grid(True)
plt.show()

# 2. Bar Chart: Salary by Employee
plt.figure(figsize=(9, 5))
plt.bar(df['Name'], df['Salary'])
plt.title("Salary by Employee")
plt.xlabel("Employee Name")
plt.ylabel("Salary")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 3. Histogram: Employee Age Distribution
plt.figure(figsize=(8, 5))
plt.hist(df['Age'], bins=6, edgecolor='black')
plt.title("Distribution of Employee Ages")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
#week - 8

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the employee dataset
df = pd.read_csv("employee.csv")

print("Employee Dataset:")
print(df.head())

# Select numerical columns
numeric_df = df.select_dtypes(include="number")

# 1. Correlation Heatmap
plt.figure(figsize=(8, 6))

correlation = numeric_df.corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Employee Data Correlation Heatmap")
plt.tight_layout()
plt.show()

# 2. Pairplot
sns.pairplot(
    df,
    vars=["Age", "Salary", "Experience"],
    diag_kind="hist"
)

plt.suptitle("Pairplot of Employee Data", y=1.02)
plt.show()
```

```python
#week - 9

import pandas as pd
import plotly.express as px

# Load employee dataset
df = pd.read_csv("employee.csv")

print("Employee Dataset:")
print(df.head())

# 1. Interactive Bar Chart: Salary by Employee
fig1 = px.bar(
    df,
    x="Name",
    y="Salary",
    color="Department",
    hover_data=["Emp_ID", "Age", "Experience"],
    title="Employee Salary Comparison",
    labels={"Salary": "Salary", "Name": "Employee Name"}
)
fig1.show()

# 2. Interactive Scatter Plot: Experience vs Salary
fig2 = px.scatter(
    df,
    x="Experience",
    y="Salary",
    color="Department",
    size="Age",
    hover_name="Name",
    hover_data=["Emp_ID"],
    title="Employee Experience vs Salary"
)
fig2.show()

# 3. Interactive Histogram: Age Distribution
fig3 = px.histogram(
    df,
    x="Age",
    color="Department",
    nbins=10,
    title="Employee Age Distribution",
    labels={"Age": "Employee Age"}
)
fig3.show()

# 4. Interactive Box Plot: Salary Distribution
fig4 = px.box(
    df,
    x="Department",
    y="Salary",
    color="Department",
    hover_data=["Name", "Experience"],
    title="Salary Distribution by Department"
)
fig4.show()
```
